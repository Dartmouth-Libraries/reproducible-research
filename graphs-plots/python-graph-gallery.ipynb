```python
# library & dataset
import seaborn as sns
df = sns.load_dataset('iris')

# Plot the histogram thanks to the displot function
# this uses the kde - KDE represents the data using a continuous probability density curve 
sns.displot( data=df["sepal_length"], kde=True )
# current kernel: python 3.12
```

```python
# bin number 
# libraries & dataset
import seaborn as sns
import matplotlib.pyplot as plt
  
sns.set_theme(style="darkgrid")
df = sns.load_dataset("iris")

sns.histplot(data=df, x="sepal_length", bins=20)
plt.show()
```

```python
# libraries & dataset
import seaborn as sns
import matplotlib.pyplot as plt
from pypalettes import load_palette

sns.set_theme(style="darkgrid")
df = sns.load_dataset("iris")
palette = load_palette("Acadia", keep_first_n=3)

sns.boxplot(hue=df["species"], y=df["sepal_length"], palette=palette)
plt.show()
```
