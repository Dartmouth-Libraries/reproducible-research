```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1️⃣ Load data ------------------------------------------------
url = ("https://raw.githubusercontent.com/holtzy/The-Python-Graph-"
       "Gallery/master/static/data/mtcars.csv")
df = pd.read_csv(url)

# Compute group‑wise means, keep only columns (mpg = miles per gallon, wt = weight)
group_means = (
    df.groupby("cyl")[["mpg", "wt"]]
      .mean()
      .rename(columns={"mpg": "Avg MPG", "wt": "Avg Weight (1000 lbs)"})
)

# Reshape for heat‑map (cylinders as rows, metrics as columns, index is cylinder count (4, 6, 8)
heat_data = group_means

# Plot - diverging palette so that the two metrics are visually distinct
plt.figure(figsize=(5, 2.5))

#  
cmap = {"Avg MPG": "YlGn",          # greener = more mpg (better)
        "Avg Weight (1000 lbs)": "OrRd"}  # redder = heavier

#   seaborn can plot a single DataFrame; we give it a custom colormap
sns.heatmap(
    heat_data,
    annot=True, fmt=".2f",              # show numbers with two decimals
    cmap="YlGnBu",                      # overall colour map (still works)
    cbar_kws={"label": "Scaled value"},
    linewidths=.5,
    linecolor="grey"
)

plt.title("Average MPG & Weight per Cylinder Count")
plt.ylabel("Cylinders")
plt.tight_layout()
plt.show()
```

```python

#  Simple heat‑map: avg mpg & avg weight per cylinder group

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = ("https://raw.githubusercontent.com/holtzy/The-Python-Graph-"
       "Gallery/master/static/data/mtcars.csv")
df = pd.read_csv(url)

# Compute group‑wise means keep only the columns we care about (mpg = miles per gallon, wt = weight)
group_means = (
    df.groupby("cyl")[["mpg", "wt"]]
      .mean()
      .rename(columns={"mpg": "Avg MPG", "wt": "Avg Weight (1000 lbs)"})
)

# Reshape for heat‑map (cylinders as rows, metrics as columns)
#    The index will be the cylinder count (4, 6, 8)
heat_data = group_means

# Plot -----------------------
plt.figure(figsize=(5, 2.5))

#   • Use a diverging palette so that the two metrics are visually distinct
#   • vmax / vmin are set per‑metric to keep the colour‑scales comparable
cmap = {"Avg MPG": "YlGn",          # greener = more mpg (better)
        "Avg Weight (1000 lbs)": "OrRd"}  # redder = heavier

#   seaborn can plot a single DataFrame; we give it a custom colormap
sns.heatmap(
    heat_data,
    annot=True, fmt=".2f",              # show numbers with two decimals
    cmap="YlGnBu",                      # overall colour map (still works)
    cbar_kws={"label": "Scaled value"},
    linewidths=.5,
    linecolor="grey"
)

plt.title("Average MPG & Weight per Cylinder Count")
plt.ylabel("Cylinders")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

from datetime import datetime, timedelta
#Dataset
#Let's consider a dataset with 3 variables:

#date
#temperature: the first series to display. Ranges from 0 to 10.
#price: the second series to display. Ranges from 20 to 120.
# Ensures reproducibility of random numbers
rng = np.random.default_rng(1234)

# timedelta(i) adds "i" days to the 1st of January of 2019
date = [datetime(2019, 1, 1) + timedelta(i) for i in range(100)]
temperature = np.arange(100) ** 2.5 / 10000 + rng.uniform(size=100) 
price = np.arange(120, 20, -1) ** 1.5 / 10 + rng.uniform(size=100)
```

```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-

"""
Load penguin measurements (bill length, bill depth, body mass) for
three species and plot body mass using a different colour per species.
"""

# import libraries 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# raw data, 10 measurements per species of penguin
# Adelie, Chinstrap, Gentoo 
bill_length_mm = [
    [39.0, 37.7, 33.5, 35.3, 37.6, 42.9, 41.4, 34.0, 33.1, 36.0],  # Adelie
    [50.6, 45.4, 50.5, 50.8, 46.9, 50.8, 51.3, 52.2, 49.2, 51.5],  # Chinstrap
    [47.5, 46.1, 50.0, 50.0, 43.3, 48.5, 49.4, 48.6, 47.5, 44.5],  # Gentoo
]

bill_depth_mm = [
    [18.7, 16.0, 19.0, 18.9, 19.1, 17.6, 18.6, 17.1, 16.1, 17.8],
    [19.4, 18.7, 18.4, 19.0, 16.6, 18.5, 19.2, 18.8, 18.2, 18.7],
    [14.2, 15.1, 15.3, 15.9, 13.4, 15.0, 15.8, 16.0, 14.0, 14.7],
]

body_mass_g = [
    [3650.0, 3075.0, 3600.0, 3800.0, 3750.0, 4700.0, 3700.0, 3400.0, 2900.0, 3450.0],
    [3800.0, 3525.0, 3400.0, 4100.0, 2700.0, 4450.0, 3650.0, 3450.0, 4400.0, 3250.0],
    [4600.0, 5100.0, 5550.0, 5350.0, 4400.0, 4850.0, 4925.0, 5800.0, 4875.0, 4850.0],
]

# build a list of the species 
species = (["Adelie"] * 10) + (["Chinstrap"] * 10) + (["Gentoo"] * 10)

# flatten lists 
def flatten(nested):
    """Convert a list‑of‑lists (3 × 10) → 1‑D list of length 30."""
    return [value for sublist in nested for value in sublist]

bill_length = flatten(bill_length_mm)
bill_depth   = flatten(bill_depth_mm)
body_mass    = flatten(body_mass_g)

# build data frame 
df = pd.DataFrame({
    "species":      species,
    "bill_length_mm": bill_length,
    "bill_depth_mm":  bill_depth,
    "body_mass_g":    body_mass,
})

print("First five rows of the DataFrame")
print(df.head())
print("\nDataFrame shape:", df.shape)

# plot: body‑mass by species (different colour per species)
plt.figure(figsize=(8, 5))

# Seaborn’s stripplot shows every individual observation.
sns.stripplot(
    data=df,
    x="species",
    y="body_mass_g",
    jitter=True,                 # spreads points a little for readability
    palette=["#1f77b4", "#ff7f0e", "#2ca02c"],  # blue, orange, green
    size=8,
    edgecolor="gray",
    linewidth=0.5,
)

plt.title("Penguin Body Mass by Species")
plt.ylabel("Body mass (g)")
plt.xlabel("Species")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-

"""
Penguin Visualization Tutorial
--------------------------------
A single‑file, fully‑commented script that:

 Loads the Palmer‑Penguins dataset directly from Seaborn.
 Cleans the data (drops rows with missing values for the variables we use).
 Produces four hands‑on plots that highlight differences among the
   three species (Adelie, Chinstrap, Gentoo):
   • Boxplot + stripplot of body mass
   • Split violin plot of bill length by sex
   • PairGrid (scatter‑matrix) of key morphometrics
   • FacetGrid of flipper length vs. body mass
 (Optional) Correlation‑matrix heatmap.
 Shows how to export the figures to PNG/PDF.

The script is written so that each block can be copied into a Jupyter
notebook cell if you prefer an .ipynb workflow.
"""

# Imports & global style settings
# 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Use a clean white‑grid style for all figures
sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 100   # nicer on most monitors

# Load the Palmer‑Penguins dataset
# 
# Seaborn ships a tidy copy of the dataset (same as the R version)
penguins = sns.load_dataset("penguins")

# Quick look at the raw data
print("\n=== First rows of the raw dataset ===")
print(penguins.head())
print("\n=== DataFrame info ===")
print(penguins.info())

# Clean the data
# For the tutorial we need complete rows for the numeric traits
# and the column `sex`.  Drop rows that are missing any of these.
cols_needed = ["bill_length_mm", "bill_depth_mm",
               "flipper_length_mm", "body_mass_g", "sex"]
penguins_clean = penguins.dropna(subset=cols_needed)

print(f"\nCleaned dataset shape: {penguins_clean.shape}")   # should be 333×7 (as of v0.1.4)

# Plot 1 – Boxplot of Body Mass by Species
# ----------------------------------------------------
plt.figure(figsize=(7, 5))
sns.boxplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    palette="Set2",          # three pastel colours
    notch=True,
    linewidth=1.2
)

# Overlay each observation so the viewer sees the raw points
sns.stripplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    color="black",
    size=3,
    jitter=True,
    alpha=0.5
)

plt.title("Penguin Body Mass by Species")
plt.ylabel("Body mass (g)")
plt.xlabel("")
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Plot 2 – Split Violin of Bill Length (by Sex)
# ----------------------------------------------------
plt.figure(figsize=(8, 5))
sns.violinplot(
    data=penguins_clean,
    x="species",
    y="bill_length_mm",
    hue="sex",
    split=True,               # left = female, right = male
    palette="Pastel1",
    inner="quartile"
)

plt.title("Bill Length Distribution by Species & Sex")
plt.ylabel("Bill length (mm)")
plt.xlabel("")
plt.legend(title="Sex", loc="upper right")
plt.tight_layout()
plt.show()

# Plot 3 – PairGrid (scatter‑matrix) of Morphometrics
# ----------------------------------------------------
num_vars = ["bill_length_mm", "bill_depth_mm",
            "flipper_length_mm", "body_mass_g"]

g = sns.PairGrid(
    penguins_clean,
    vars=num_vars,
    hue="species",
    palette="Dark2",
    height=2.2
)

# Upper triangle – regression line + scatter
g.map_upper(
    sns.regplot,
    scatter_kws={"s": 30, "alpha": 0.6},
    truncate=False
)

# Lower triangle – plain scatter
g.map_lower(sns.scatterplot, s=30, alpha=0.7)

# Diagonal – KDE of each variable
g.map_diag(sns.kdeplot, fill=True, alpha=0.6)

g.add_legend(title="Species")
plt.suptitle("Morphological Relationships Across Species", y=1.02)
plt.show()

# Plot 4 – FacetGrid of Flipper Length vs Body Mass (by Species)
# ----------------------------------------------------
facet = sns.FacetGrid(
    penguins_clean,
    col="species",
    hue="sex",
    palette="Set1",
    height=4,
    aspect=1
)

# Scatter points (one colour per sex)
facet.map(
    sns.scatterplot,
    "flipper_length_mm", "body_mass_g",
    s=60, alpha=0.7, edgecolor="white"
)

# Add a regression line per facet
facet.map(
    sns.regplot,
    "flipper_length_mm", "body_mass_g",
    scatter=False, truncate=False, ci=None, lw=2
)

facet.add_legend(title="Sex")
facet.set_axis_labels("Flipper length (mm)", "Body mass (g)")
facet.set_titles(col_template="{col_name} penguins")
facet.fig.suptitle("Flipper Length ↔ Body Mass by Species", y=1.05)
plt.tight_layout()
plt.show()

# -(Optional) Plot 5 – Correlation Matrix Heatmap
# ----------------------------------------------------
corr = penguins_clean[num_vars].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1, vmax=1,
    linewidths=.5,
    linecolor="gray"
)
plt.title("Correlation Matrix of Morphological Traits")
plt.tight_layout()
plt.show()

# - Export the last figure (FacetGrid) as PNG and PDF
# ----------------------------------------------------
# The FacetGrid object `facet` is still in memory after the show()
# call, so we can save it directly.
output_base = "penguins_flipper_bodymass"
facet.savefig(f"{output_base}.png", dpi=300, bbox_inches="tight")
facet.savefig(f"{output_base}.pdf", bbox_inches="tight")
print(f"\nSaved FacetGrid as {output_base}.png and {output_base}.pdf")

# ----------------------------------------------------
# End of script
# ----------------------------------------------------
print("\n=== Tutorial finished! ===")
```

```python
# ## 1) Scatter with semantic mappings
#Goal: Show numeric-numeric relationship while encoding two categorical variables (species -> color, sex -> marker).

plt.figure(figsize=(6.5, 4.5))
ax = sns.scatterplot(
    data=penguins,
    x="bill_length_mm", y="bill_depth_mm",
    hue="species", style="sex", s=80, edgecolor="black", linewidth=0.5
)
ax.set(
    title="Bill length vs depth by species and sex",
    xlabel="Bill length (mm)", ylabel="Bill depth (mm)"
)
sns.move_legend(ax, "upper right", frameon=True, title="Species / Sex")
plt.tight_layout()
plt.show()
```

```python


## 2) Violin + inner boxplot
#Goal: Compare distributions across categories while retaining a robust summary (box). Note: cut=0 trims KDE tails to observed range.

plt.figure(figsize=(6.5, 4.2))
ax = sns.violinplot(
    data=penguins,
    x="species", y="flipper_length_mm",
    hue="species", dodge=False, palette="Set2", cut=0
)
sns.boxplot(
    data=penguins,
    x="species", y="flipper_length_mm",
    whis=(5, 95), width=0.2, showcaps=False, boxprops={"facecolor":"white"},
    showfliers=False
)
ax.set(
    title="Flipper length distribution by species",
    xlabel="Species", ylabel="Flipper length (mm)"
)
plt.tight_layout()
plt.show()
```

```python
# ## 3) Histogram + KDE (faceted by species)
#Goal: Univariate distributions with consistent binning; facet by species to avoid overplotting.

bins = np.arange(2500, 6500, 250)
g = sns.displot(
    data=penguins,
    x="body_mass_g", col="species",
    bins=bins, kde=True, color="#4C72B0", height=3.2, aspect=1
)
g.set_axis_labels("Body mass (g)", "Count")
g.set_titles("{col_name}")
plt.suptitle("Body mass distributions by species", y=1.05)
plt.tight_layout()
plt.show()
```

```python
# ## 4) Correlation heatmap (numeric features)
#Goal: Summarize linear relationships across measurements. Use diverging palette and annotation for readability.

num_cols = ["bill_length_mm","bill_depth_mm","flipper_length_mm","body_mass_g"]
corr = penguins[num_cols].corr(method="pearson")

plt.figure(figsize=(5.8, 4.6))
ax = sns.heatmap(
    corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
    square=True, cbar_kws={"label": "Pearson r"},
    linewidths=.5, linecolor="white"
)
ax.set_title("Correlation among penguin measurements")
plt.tight_layout()
plt.show()
```

```python
## 5) Regression with confidence interval (faceted)
# and facets (multiple graphs side-by-side)
# Goal: Visualize the flipper-length vs body-mass relationship with linear fit and 95% CI, separated by species.
g = sns.lmplot(
    data=penguins,
    x="flipper_length_mm", y="body_mass_g",
    col="species", height=3.2, aspect=1,
    scatter_kws=dict(s=40, alpha=.8, edgecolor="black"),
    line_kws=dict(color="crimson", linewidth=2)
)
g.set_axis_labels("Flipper length (mm)", "Body mass (g)")
g.set_titles("{col_name}")
plt.suptitle("Body mass vs flipper length with linear fit", y=1.05)
plt.tight_layout()
plt.show()
```

```python
# ## 6) Categorical bar with error bars (mean +/- 95% CI)
# Goal: Show group summaries and uncertainty; sex used as dodge grouping.

plt.figure(figsize=(6.8, 4.4))
ax = sns.barplot(
    data=penguins,
    x="species", y="bill_length_mm",
    hue="sex", errorbar="ci", estimator=np.mean, n_boot=1000,
    palette="muted", capsize=.15
)
ax.set(
    title="Mean bill length by species and sex (+/-95% CI)",
    xlabel="Species", ylabel="Mean bill length (mm)"
)
sns.move_legend(ax, "upper left", frameon=True)
plt.tight_layout()
plt.show()
```
